```python
for col in dataset.columns:
    if(dataset[col].dtypes == 'O'):
        qual.append(col)
    else:
        quan.append(col)

print(qual)
print(quan)
```

```python
def get_Quan_Qual(source):
    qual=[]
    quan=[]
    for col in source.columns:
        if(source[col].dtypes == 'O'):
            qual.append(col)
        else:
            quan.append(col)

    return qual,quan
    
```

```python
x,y=get_Quan_Qual(dataset)
print(x)
print(y)
```
