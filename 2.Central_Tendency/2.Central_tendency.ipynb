# Outlier capping or Winsorization or flooring and capping (Replacing outlier value)

Outlier capping (also known as Winsorization or flooring and capping) is a technique used to handle extreme values by restricting them to a predefined maximum and minimum threshold. Instead of removing rows entirely (trimming), capping alters the outlier values to match your acceptable limits.

Below code uses chained indexing (e.g., dataset[col][condition] = value)

```

for columnName in lesser:
    dataset[columnName][dataset[columnName] < descriptive[columnName]["Lesser"]] = descriptive[columnName]["Lesser"]

```

In this specific code, the rows are not actually looped one by one.
Instead of using a for loop to check each row individually, pandas uses a highly optimized technique called vectorization and boolean indexing.
Here is exactly how pandas processes the rows behind the scenes without explicitly looping through them:

**Step 1: Creating a Mask for All Rows**

When pandas evaluates the inner bracket condition—for example, dataset[columnName] < limit—it looks at the entire column at once. It compares the limit against every single row simultaneously in memory using fast C-based code.
This generates a series of True and False values (called a boolean mask) corresponding to each row:
• True means the row's value is an outlier and needs to be changed.
• False means the row's value is fine and should be left alone.

**Step 2: Filtering and Updating**

When you pass that mask back into the dataset (dataset[columnName][ mask ] = limit), pandas uses the True positions to pinpoint exactly which rows to overwrite. It completely skips the rows marked False.


 **Why is Chained Indexing a Problem?**
 
Chained indexing causes two main issues: unpredictable behavior and performance warnings.
When pandas extracts the column in the first step, it does not guarantee whether it is handing you a view (a direct reference back to the original memory block) or a copy (a brand new, isolated piece of data in memory).
• If it creates a copy: You end up modifying a temporary, standalone Series object that immediately disappears. The original dataset DataFrame remains completely unchanged.
• If it creates a view: The modification successfully trickles down and changes your original dataset.
Because pandas cannot always predict which one it will do, it will throw the infamous warning:
SettingWithCopyWarning: A value is trying to be set on a copy of a slice from a DataFrame.

**outlier replacement can be done using  Label-based Indexing (.loc)**

dataset.loc[condition, columnName] = value

# Frequency, Relative Frequence, Cumulative Frequency (cumsum)
* **Frequency:** The exact count of how many times a specific value, category, or range appears in a dataset
* **Relative Frequency:** The proportion or percentage of times a value occurs relative to the total number of observations , calculated as ({Frequency} / n). It allows data scientists to compare categories or datasets of different sizes
* **Cumulative Frequency:** The running total of frequencies as you move down an ordered list of categories or numerical intervals. It shows how many observations fall at or below a certain threshold.




**How Data Scientists Use This in Practice**

* 1. Frequency (\(f\)) tells you raw volume: You see that 20 tickets took between 5 and 10 minutes. This helps management know where the highest operational load is concentrated.
* 2. Relative Frequency tells you proportion/probability: You learn that \(40\%\) of tickets fall into the 5–10 minute bucket, while only \(10\%\) take longer than 15 minutes. This scales the metric so you can compare it against a dataset of 1,000 tickets next month.
* 3. Cumulative Frequency tells you threshold coverage: Looking at the cumulative column, you see that 45 out of 50 tickets (or \(90\%\)) are resolved within 15 minutes. If your business SLA mandates that \(90\%\) of customers must be helped within 15 minutes, this table immediately proves you are just hitting the threshold.
