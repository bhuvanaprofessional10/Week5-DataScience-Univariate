```python
def get_Quan_Qual(source):
    ql=[]
    qn=[]
    for col in source.columns:
        if(source[col].dtypes == 'O'):
            ql.append(col)
        else:
            qn.append(col)

    return ql,qn
    
```

```python
qual,quan=get_Quan_Qual(dataset)
print(qual)
print(quan)
```

```python
import numpy as np

UserTable=pd.DataFrame(index=["Mean","Median","Mode","Minimum","Q1:25%","Q2:50%","Q3:75%","99%","Q4:100%"],columns=quan)
for col in quan:
    UserTable[col]["Mean"]=float(dataset[col].mean())
    UserTable[col]["Median"]=float(dataset[col].median())    
    UserTable[col]["Mode"]=float(dataset[col].mode()[0])
    UserTable[col]["Minimum"]=float(dataset[col].min())
    UserTable[col]["Q1:25%"]=dataset.describe()[col]["25%"]
    UserTable[col]["Q2:50%"]=dataset.describe()[col]["50%"]
    UserTable[col]["Q3:75%"]=dataset.describe()[col]["75%"]
    UserTable[col]["99%"]=float(np.percentile(dataset[col],99))
    UserTable[col]["Q4:100%"]=dataset.describe()[col]["max"]
UserTable
```
